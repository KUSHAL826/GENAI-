```python
from google import genai
from google.colab import userdata
from google.genai import types
import numpy as np

client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
EMB_MODEL="gemini-embedding-001"
EMB_DIM=768
```

```python
def embed(text):
  response=client.models.embed_content(
      model=EMB_MODEL,
      contents=text,
      config=types.EmbedContentConfig(
          output_dimensionality=EMB_DIM
      )
  )
  return response.embeddings[0].values
def cosine_sim(a,b):
  a=np.array(a)
  b=np.array(b)

  return (np.dot(a,b))/(np.linalg.norm(a)*(np.linalg.norm(b)))

movies = [
    "Interstellar - A scientist travels through space to save humanity.",
    "Inception - A team enters people's dreams to perform a difficult mission.",
    "The Martian - An astronaut becomes stranded on Mars and uses science to survive.",
    "Avengers Endgame - Superheroes work together to defeat a powerful enemy and save the universe.",
    "Jurassic Park - Scientists create dinosaurs using genetic technology, but the creatures escape.",
    "Finding Nemo - A father travels across the ocean to find his lost son.",
    "Toy Story - Animated toys come to life and go on adventures about friendship.",
    "The Dark Knight - A superhero battles a dangerous criminal who creates chaos in a city."
]
embeddings=np.array([
    embed(movie)
    for movie in movies
])
user_input=input("Enter movie type you need to watch: ")
user_embed=embed(user_input)
similarity=[]
for i,embedding in enumerate(embeddings):
  similarity.append((i,cosine_sim(user_embed,embedding)))

best_index,best_score=max(similarity,key=lambda x:x[1])
print("RECOMENDATIONS...........")
if best_score<0.3:
  print("NO MOVIES RECOMMENDED")
else:
  print(f"MOVIE:{movies[best_index]}")
  print(f"MATCHING SCORE:{best_score:0.4f}")
```
