```python
from google import genai
from google.colab import userdata
from google.genai import types

client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
MODEL="gemini-3.5-flash-lite"
```

```python
user_input=input("Enter the prompt: ")
prompt=f"""

You will be given with user prompt:{user_input}, so optimise and improvise with the userprompt to get better answer.
"""

response=client.models.generate_content(
    model=MODEL,
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=0.5,
        max_output_tokens=1024,
        system_instruction="You are Senior Prompt Engineer,reply with Hi, I am Your Prompt Optimiser.",
        thinking_config=types.ThinkingConfig(thinking_level='high')
    )
)
print(response.text)
```
