```python
from google import genai
from google.colab import userdata
from google.genai import types
import numpy as np
client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
EMB_MODEL="gemini-embedding-001"
EMB_DIM=768
```

```python
def embed(text):
  response=client.models.embed_content(
      model=EMB_MODEL,
      contents=text,
      config=types.EmbedContentConfig(
          output_dimensionality=768
      )
  )
  return response.embeddings[0].values

def cosine_sim(a,b):
  a=np.array(a)
  b=np.array(b)

  return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))

job_role=input("Enter the Vacant job role: ")
job_embed=embed(job_role)
n=int(input("Enter number of candidates applied: "))
sentences=[]
for i in range(n):
  sentence=input(f"Enter skills of person{i+1}: ")
  sentences.append(sentence)

embeddings=np.array([
    embed(sentence)
    for sentence in sentences
])

similarity=[]
for i,embedding in enumerate(embeddings):
  score=cosine_sim(job_embed,embeddings[i])
  similarity.append((i,score))
best_index,best_score=max(similarity,key=lambda x:x[1])

print("BEST MATCH")
print(f"JOB ROLE APPLIED: {sentences[best_index]}")
print(f"BEST SCORE: {best_score:.4f}")


#printing in order:
similarity.sort(key=lambda x:x[1], reverse=True)
print("ORDER: ")
for i,(index,score) in enumerate(similarity):
  print(f"{i+1}.{sentences[index]}-{score}")
```
