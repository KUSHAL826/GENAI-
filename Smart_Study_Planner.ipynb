```python
from google import genai
from google.colab import userdata
from google.genai import types
import json
from typing import Optional
from pydantic import BaseModel

client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
MODEL="gemini-3.5-flash-lite"
```

```python
user_input=input("Enter your name,topics to study,days avalaible,number of hours available on each day,level")
prompt=f"""
You are Senior Study timetable planner for students.
Consider the Information provided by Student:{user_input}.
Based on topics,days available,and no of hours available on each day and level, plan the routine foe student.
Return JSON format with keys( name(string),routine_of_that(list containing subject,number of hours(even per day 1ormore subjects can be planned)))
"""

response=client.models.generate_content(
    model=MODEL,
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=0.8,
        max_output_tokens=1500,
        response_mime_type="application/json",
        system_instruction="You are study planner, Reply (I am Your study planner)",
        thinking_config=types.ThinkingConfig(thinking_level='low')
    )
)
data=json.loads(response.text)
print(json.dumps(data,indent=2))
```
