```python
import numpy as np
arr1=np.array([1,2,3])
print("1D array: ",arr1)

arr2=np.arange(1,10).reshape(3,3)
print("2D array: ",arr2)

arr3=np.arange(1,28).reshape(3,3,3)
print("3D array: ",arr3)

print(arr3[1,1,1])
```

```python
arr1=np.array([[1,2,3],[4,5,6]],dtype=np.int64)
print(arr1)

arr=np.ones((3,3),dtype=np.int32)
print(arr)
arr2=np.identity(2,dtype=np.int32)
print(arr2)

#Linspace
arr1=np.linspace(-10,10,5)
print(arr1)
print(arr1.astype("int32"))
```

```python
arr1=np.array([[1,2,3],[4,5,6]],dtype=np.int64)
print(np.max(arr1,axis=0))
print(np.max(arr1,axis=1))
#np.min()
#np.sum
#np.prod
#np.diff
```

```python
#round ceil floor

arr=np.array([[1.2,3.9,4.5],[4.6,6.5,7.9]])
print(arr.round().astype("int32"))

print(np.ceil(arr).astype("int32"))

print(np.floor(arr).astype("int32"))
```

```python
# nD -> 1D

arr=np.arange(1,17).reshape(4,4)
print(arr)
arr=np.ravel(arr)
print(arr)
print(arr.size)
```
