```python
from google import genai
from google.colab import userdata
from google.genai import types

client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
MODEL='gemini-3.5-flash-lite'
```

```python
system_instruction="""
                      You are Senior code explainer Reply whenever new session starts with Hi,I am your code Assistant.
                      You will be given with code/snippet, then explain the code and if any errors/sytax error correct error with explination showing wrong code and corrected code,concepts used,further improvements.
                      first identify in which language code is given.
                      then reply to questions asked by user
                      FIRST LINE PRINT LANGUAGE OF CODE/SNIPPET GIVEN
                  """
chat=client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        temperature=0.8,
        max_output_tokens=1024,
        system_instruction=system_instruction,
        thinking_config=types.ThinkingConfig(thinking_level='medium')
    ),
    history=[]
)

print("ENTER THE CODE!!!!\n\n")
while True:
  user_input=input("CODE ASSITANT: Enter the code or question (enter exit,bye,quit to terminate session): ")
  if user_input.lower() in ['exit','bye','quit']:
    print("THANK YOU\nEXITING...........\nEXITED")
    break
  response=chat.send_message(user_input)
  print(response.text)
```
