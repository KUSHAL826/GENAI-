```python
from google import genai
from google.colab import userdata
from google.genai import types

client = genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
MODEL="gemini-3.5-flash-lite"
#start fresh session
chat=client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        temperature=0.8,
        max_output_tokens=1500,
        system_instruction="Start with hello i am your assistant",
        thinking_config=types.ThinkingConfig(thinking_level='low')
    ),
    history=[]
)
print("BOT IS READY")

while True:
  user_input=input("YOU: ")
  if user_input.lower() in ['bye','exit','quit']:
    break
  else:
    response=chat.send_message(user_input)
    print(f"BOT:{response.text}")
```

```python
print("-----------HISTORY-------------")
for i,msg in enumerate(chat.get_history()):
  role = 'USER' if msg.role=='user' else 'BOT'
  print(f"{i+1},{role}:{msg}")
```

**CHATBOT DOES:**

**1.accept user messsage**
**2.remember prior message in same conversation**
**3.generate context relevant reply**
