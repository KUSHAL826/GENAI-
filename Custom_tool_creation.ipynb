```python
from langchain_core.tools import tool

#tool method ---- method 1
@tool
def multiply_numbers(a: int, b: int) -> int:
  """
  This tool is to multiply 2numbers provided
  """
  return a * b

result = multiply_numbers.invoke({"a": 2, "b": 3})
print("PRODUCT OF 2numbers: ", result)

print("NAME OF TOOL: ", multiply_numbers.name)
print("DESCRIPTION: ", multiply_numbers.description)
print("ARGUMENTS: ", multiply_numbers.args)
```

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field

class MultiplyInput(BaseModel):
  a: int = Field(description="First number")
  b: int = Field(description="Second number")

def multiply_func(a: int, b: int) -> int:
  return a * b

multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    description="this tool is to multiply 2numbers",
    name="multiple",
    args_schema=MultiplyInput
)
print(multiply_tool.invoke({'a': 2, 'b': 3}))
```
