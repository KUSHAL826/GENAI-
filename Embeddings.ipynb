```python
from google import genai
from google.colab import userdata
from google.genai import types
import numpy as np
client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
EMBEDDING_MODELS="gemini-embedding-001"
EMBEDDING_DIMENSIONS=768
```

```python
def embed(text):
  response=client.models.embed_content(
      model=EMBEDDING_MODELS,
      contents=text,
      config=types.EmbedContentConfig(
          output_dimensionality=EMBEDDING_DIMENSIONS
      )
  )
  return response.embeddings[0].values

text=["REACT DEVELOPER WITH PAYMENT OPTIONS","REACT DEVELOPER"]
embedding=embed(text)

print(type(embedding))
print(len(embedding))
print("FIRST FEW VALUES OF EMBEDDINGS: ",embedding[0:9])
```

```python
sentences = [
    "React developer with payment gateway experience.",
    "Built UPI checkout flow for Razorpay using Next.js and Node",
    "Cardiologist with 12 years experience at AIIMS Delhi",
]

embeddings=np.array([
    embed(sentence)
    for sentence in sentences
])
print(embeddings[:9])
print("EMBEDDING SHAPE: ",embeddings.shape)
```

```python
def cosine_sim(a,b):
  a=np.array(a)
  b=np.array(b)

  return np.dot(a,b)/ (np.linalg.norm(a)*np.linalg.norm(b))

print("COSINE SIMILARITY")
print(f"React vs Node developer: {cosine_sim(embeddings[0],embeddings[1]):.4f}")
print(f"Node vs Doctor: {cosine_sim(embeddings[1],embeddings[2]):.4f}")
print(f"DOCTOR vs React: {cosine_sim(embeddings[2],embeddings[0]):.4f}")
```
