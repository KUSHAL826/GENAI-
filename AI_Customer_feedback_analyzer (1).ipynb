```python
from google import genai
from google.colab import userdata
from google.genai import types


client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
MODEL="gemini-3.5-flash-lite"
```

```python
user_input=input("\nEnter multiple costumer reviews: ")
prompt=f"""
Go through the reviews provided by multiple costumers:{user_input}
Then detect positive/negative/neutral sentiments.
extract common complaints
identify frequently requested feature
generate business response
At first give generate costumer rating number out of 10 (example 8/10)based on the reviews.(no skipping of rating)
Return in Well Formated Text Structure
"""

response=client.models.generate_content(
    model=MODEL,
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=0.6,
        max_output_tokens=1500,
        system_instruction="You senior marketing officer, Reply with(Hi,I am your marketing assistant)",
        thinking_config=types.ThinkingConfig(thinking_level='high')
    )

)

print(response.text)
```
