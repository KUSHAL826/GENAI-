```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEYS'))
```

```python
for m in client.models.list():
  print(m.name)
```

```python
response = client.models.generate_content(
    model = 'gemini-3.5-flash-lite',
    contents ='tell me one line of joke about developers.',
    config=types.GenerateContentConfig(
        system_instruction='you are a helpful assistant',
        temperature=0.5,
        max_output_tokens=200,
        thinking_config=types.ThinkingConfig(thinking_level='low')
        )
)


print("text generated by AI:",response.text)

print()
u=response.usage_metadata
print("total token count",u.total_token_count)
print("usage", response.usage_metadata)
print()
print("Prompt token count",u.prompt_token_count)
print
```
