```python
import numpy as np
arr1 = np.arange(1,11)
print(arr1)
print(arr1.shape)
print(arr1.size)
print(arr1.ndim)
print(arr1.dtype)
```

```python
arr2 = np.arange(1,10).reshape(3,3)
print(arr2)
print(arr2.shape)
print(arr2.size)
print(arr2.ndim)
print(arr2.dtype)
```

```python
arr3=np.array([10,20,30,40,50])
print(arr3)
print(arr3.shape)
print(arr3.size)
print(arr3.ndim)
print(arr3.dtype)
```

```python
#Mathematical Operations
A=np.array([10,20,30,40])
B=np.array([1,2,3,4])
print("ADDITION: ",A+B)
print("SUBTRACTION: ",A-B)
print("MULTIPLICATION: ",A*B)
print("DIVISION: ",(A/B))
print("EXPONENTIAL: ",A**B)
```

```python
#Methematical functions
values = np.array([2,4,6,8,10])
print("Square Root Of Each Element: ",np.sqrt(values))
print("Exponential Of Each Element: ",np.exp(values))
print("Logarithm Of Each Element: ",np.log(values))
print("Sum of each Element: ",np.sum(values))
print("Cumulative Sum of each element: ",np.cumsum(values))
```

```python
#Aggregate Operations
values = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print("Row Wise Sum: ",np.sum(values,axis=1))
print("Column Wise Sum: ",np.sum(values,axis=0))
print("Minimum Element: ",np.min(values))
print("Overall Mean: ",np.mean(values))
```

```python
#Statistical Functions
marks = np.array([78,85,90,66,72,88,95,60])
mean = np.mean(marks)
print("Mean: ",mean)
median = np.median(marks)
print("Median: ",median)
std = np.std(marks)
print("Standard Deviation: ",std)
var = np.var(marks)
print("Variance: ",var)
max = np.max(marks)
print("Maximum: ",max)
min = np.min(marks)
print("Minimum: ",min)
range = max - min
print("Range: ",range)
#percentile and sorting
print("25th Percentile: ",np.percentile(marks,25))
print("50th Percentile: ",np.percentile(marks,50))
print("75th Percentile: ",np.percentile(marks,75))
#average
sum = np.sum(marks)
avg = sum/(len(marks))
count = 0
for i in np.nditer(marks):
  if i > avg:
    count += 1
print("Average: ",avg)

print("Number of students above average: ",count)
```

```python
#Sales Analysis
sales = np.array([1200,1500,900,2000,1800,1700,1600])
print("Total Weekly Sales: ",np.sum(sales))
print("Average Daily Sales: ",np.average(sales))
print("Highest Daily Sales: ",np.max(sales))
print("Lowest Daily Sales: ",np.min(sales))
print("Standard Deviation Of Sales: ",np.std(sales))
days = 1
print("Days Having Sales above Average: ")
for i in np.nditer(sales):
  if i > np.average(sales):
    print("DAY: ",days)
  days+=1
```
