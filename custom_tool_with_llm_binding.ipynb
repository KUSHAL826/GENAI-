```python
import os
from google.colab import userdata
os.environ["GOOGLE_API_KEY"]=userdata.get("GEMINI_API_KEYS")
MODEL="gemini-3.5-flash-lite"
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage,ToolMessage
llm=ChatGoogleGenerativeAI(model=MODEL)


@tool
def add(a: int, b: int) -> int:
  "Tool to add to 2numbers"
  return a + b

@tool
def multiply(a: int, b: int) -> int:
  "Tool to multiply to 2numbers"
  return a * b

llm_with_tools=llm.bind_tools([add,multiply])
messages=[]
query=HumanMessage("What is sum of 2 and 10")
messages.append(query)
AI=llm_with_tools.invoke(messages)
messages.append(AI)

tool_call=AI.tool_calls[0]
print(tool_call)
tool_message=ToolMessage(contents=str(tool_call),name=tool_call["name"],tool_call_id=tool_call["id"])
messages.append(tool_message)
print(messages)

final_output=llm_with_tools.invoke(messages)
print(final_output.content[0]['text'])
```
