**1.Student Result Analyzer**
You are given the marks of a student in 5 subjects. Write a Python program that:

Stores the student's name (string) and roll number (integer) in variables.
Takes 5 subject marks as a list of floats.
Calculates the total and average marks.
Assigns a grade using the following rules:
average >= 90 → Grade A
average >= 75 → Grade B
average >= 60 → Grade C
average >= 40 → Grade D
else → Fail
Also prints which subjects the student scored below 40 in (use a loop).
Wrap all logic inside a function analyze_result(name, roll, marks).

```python
name=input("Enter name of Student")
marks=[]
total=0
print("Enter marks of 5subjects: ")
for i in range(5):
  m=int(input(f"Enter marks of subject {i+1} "))
  marks.append(m)
  total+=m
average=total/5
if(average<=0 and average>100):
  print("INVALID INPUT")
elif(average>=90 and average<=100):
  print("GRADE A")
elif(average>=75 and average<90):
  print("Grade B")
elif(average>=60 and average<75):
  print("GRADE C")
elif(average>=40 and average<60):
  print("GRADE D")
else:
  print("FAIL")

```

**2.Library Book Management System**
Problem Statement
You are building a small Library System. Implement the following using the right data structure for each requirement:

Store the library catalog as a dictionary where key = book_id (int) and value = tuple (title, author, year). (Why tuple? Because book details should be immutable.)
Maintain a list called borrowed_books to track currently issued book IDs (insertion order matters).
Maintain a set called members to store unique member IDs (no duplicates allowed).
Write the following functions:
add_book(catalog, book_id, title, author, year)
borrow_book(catalog, borrowed_books, book_id) — only if book exists & not already borrowed.
return_book(borrowed_books, book_id)
register_member(members, member_id) — should silently ignore duplicates.
show_available(catalog, borrowed_books) — print all books that are NOT in borrowed_books.
In main(), demonstrate adding 4 books, registering 3 members (try adding the same member twice), borrowing 2 books, returning 1, and finally displaying available books.

```python
def add_book(catalog,id,title,author,year):
  catalog[id]=(title,author,year)

def borrow_book(catalog, borrowed_books, book_id):
  if book_id not in catalog:
    print("BOOK ID NOT FOUND")
  elif book_id in borrowed_books:
    print("BOOKED BORROWED ALREADY")
  else:
    borrowed_books.append(book_id)
def returnbook(borrowedbooks,id):
  if id in borrowedbooks:
    borrowedbooks.remove(id)
  else:
    print("Book is not borrowed")
def register_member(members,member_id):
  members.add(member_id)
def show_available(catalog,borrowedbooks):
  for book_id,book_details in catalog.items():
    if book_id not in borrowedbooks:
      print(book_id,book_details)


catalog = {}
borrowed_books = []
members = set()

add_book(catalog, 101, "Python Basics", "John Smith", 2022)
add_book(catalog, 102, "Data Structures", "Robert Brown", 2021)

register_member(members, 1001)
register_member(members, 1002)
register_member(members, 1003)

register_member(members, 1001)



borrow_book(catalog, borrowed_books, 101)
borrow_book(catalog, borrowed_books, 103)
returnbook(borrowed_books, 101)

show_available(catalog, borrowed_books)


print("\nBorrowed Books:", borrowed_books)
print("Members:", members)
```

```python
#Q3
# Part A — Spot the Bug

def add_item(item, cart=[]):
    cart.append(item)
    return cart


print(add_item("apple"))
print(add_item("banana"))
print(add_item("milk", cart=["bread"]))
print(add_item("eggs"))


# Output:
# ['apple']
# ['apple', 'banana']
# ['bread', 'milk']
# ['apple', 'banana', 'eggs']

# Explanation:
# The default list cart=[] is created only once when the function is defined.
# Therefore, the same list is reused every time the function is called
# without providing a cart.
#
# First call adds "apple".
# Second call uses the same list, so "banana" is added after "apple".
# Third call provides its own list ["bread"], so it does not use the default list.
# Fourth call again uses the original default list, which already contains
# "apple" and "banana", so "eggs" is added to it.


# Part B — Fix It

def add_item_fixed(item, cart=None):
    # None is immutable and is safe to use as a default value.
    # A new list is created inside the function for every call where
    # no cart is provided.
    if cart is None:
        cart = []

    cart.append(item)
    return cart


print("\nFixed Function:")
print(add_item_fixed("apple"))
print(add_item_fixed("banana"))
print(add_item_fixed("milk", cart=["bread"]))
print(add_item_fixed("eggs"))

# Output:
# ['apple']
# ['banana']
# ['bread', 'milk']
# ['eggs']


# Part C — Build the Shopping Cart

def create_cart(owner, discount=0):
    # discount is an integer/float, which is immutable.
    # Each call creates a new dictionary and a new empty list.
    return {
        "owner": owner,
        "items": [],
        "discount": discount
    }


def add_to_cart(cart, name, price, qty=1):
    # A new dictionary representing the item is added to the cart's list.
    cart["items"].append({
        "name": name,
        "price": price,
        "qty": qty
    })


def update_price(price_tuple, new_price):
    # Tuples are immutable, so their elements cannot be changed.
    # Attempting to modify a tuple element raises TypeError.

    try:
        price_tuple[0] = new_price
    except TypeError as e:
        print("TypeError:", e)


def calculate_total(cart):
    total = 0

    # Loop through every item in the cart.
    for item in cart["items"]:
        total += item["price"] * item["qty"]

    # Apply the discount percentage.
    discount_amount = total * cart["discount"] / 100

    final_total = total - discount_amount

    return final_total


# Demonstration

# Customer 1
cart1 = create_cart("Kushal", discount=10)

add_to_cart(cart1, "Laptop", 50000, 1)
add_to_cart(cart1, "Mouse", 1000, 2)


# Customer 2
cart2 = create_cart("Gagan", discount=5)

add_to_cart(cart2, "Keyboard", 2000, 1)
add_to_cart(cart2, "Headphones", 3000, 1)


# Display both carts

print("\nCart 1:")
print(cart1)

print("\nCart 2:")
print(cart2)


# Calculate totals

print("\nTotal for", cart1["owner"], ":", calculate_total(cart1))
print("Total for", cart2["owner"], ":", calculate_total(cart2))


# Prove that the carts are independent

add_to_cart(cart1, "USB Cable", 500, 1)

print("\nAfter adding USB Cable to Cart 1:")

print("Cart 1:", cart1)
print("Cart 2:", cart2)

# Cart 2 is unchanged because each call to create_cart()
# creates a separate dictionary and a separate empty items list.


# Tuple immutability demonstration

price_tuple = ("Laptop", 50000)

print("\nOriginal tuple:", price_tuple)

update_price(price_tuple, 45000)

print("Tuple after attempted modification:", price_tuple)

# The tuple remains unchanged because tuples are immutable.
# Trying to assign a new value to price_tuple[0] raises TypeError.
```
