```python
import numpy as np
import pandas as pd
```

Pandas series is a column of data in Table format where each data is assaigned with labels called as Indexes
data:items
index:label assosciated with data
name:Userdefined name for the data
dtype:Data Type of data

```python
data1 = pd.Series(data=[10,20,30],index=["Milk","Bread","Butter"],name="Availabe Stock")
print(data1)
```

Creating series:
1:list
2:numpy array
3:dictinary
4:custom index
5:scalar value


```python
#Creating Series With List
country = ['India','China','USA','Pakistan']
data2 = pd.Series(country)
print(data2)
```

```python
batsmanruns = [48,64,128,95,82,65]
data3 = pd.Series(batsmanruns,dtype='int32')
print(data3)
```

```python
subject_names = ['Cn','TOC','ADA','SEPM','AI']
subject_marks = [33,35,36,28,35]
data4 = pd.Series(data=subject_marks,index=subject_names,name="IA ANALYSIS")
print(data4)
```

```python
#creating series with numpy array
array = np.arange(0,10)
data5 = pd.Series(array,dtype="int32")
print(data5)
```

```python
#creating Series with Dictionary
dict = {"Cn":33,"TOC":35,"ADA":36,"SEPM":28,"AI":35}
data6 = pd.Series(dict,dtype="int32")
print(data6)
```

```python
#creating series with custom index
values = [10,20,30,40]
data7 = pd.Series(data=values,index=['a','b','c','d'])
print(data7)
```

```python
#creating series with Scalar values
data8 = pd.Series(4,index=['a','b','c','d'])
print(data8)
```

Summary:

```python
#Mathematical Functions
values = np.random.randint(50000,100000,size=5000)
data = pd.Series(values)
print(data)
```

```python
print(data.sum())
print(data.product())
print(data.min())
print(data.sort_values())
print("Mean",data.mean())
print("Median: ",data.median())
print("MODE: ",data.mode())
print("Variance: ",data.var())
print("Standard deviation: ",data.std())
```
