```python
from google import genai
from google.genai import types
from google.colab import userdata

client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
```

```python
for m in client.models.list():
  print(m.name)
```

```python
def multicontent(choice):
  if(choice==1):
    prompt=f"""Act as Senior Blog writer
            Write a blog post on topic AI in Education system with their pros and cons
            Start with introduction about AI,then tell about AI impact in education system (pros and cons),how can it utilized in positive way
            blog must be 100words
            """
  elif(choice==2):
    prompt=f"""Act as Senior Email writer
              Write professional email to Client with whom our company had meeting about our SaaS service
              Start with telling that we had meeting which was very good session,then inform about our SaaS service then seek their decision about subscribing to our servive
              mail must be professional,not like spam,in correct format,not more than 100words
          """
  elif(choice==3):
    prompt=f"""Act as Senior Marketing agent
              Write instagram post caption on topic AI affect on IT professionals
              not more than 25-30words"""
  else:
    print("INVALID INPUT")
    return ""

  response=client.models.generate_content(
      model="gemini-3.5-flash-lite",
      contents=prompt,
      config=types.GenerateContentConfig(
          temperature=0.6,
          max_output_tokens=15000,
          system_instruction="Your are Helpfull assistant,start with Hi,this is your assistant,with what may i help you",
          thinking_config=types.ThinkingConfig(thinking_level='high')
      )
  )
  return response.text


ch=int(input("Enter your choice:\n1:Blog Writing on Topic AI in Education System\n2:Follow up email drafting\n3:Caption for instagram post: "))
res=multicontent(ch)
if(res==""):
  pass
else:
  print(res)
```
