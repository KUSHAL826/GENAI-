```python
from google import genai
from google.genai import types
from google.colab import userdata

client=genai.Client(api_key=userdata.get('GEMINI_API_KEYS'))
```

```python
for m in client.models.list():
  print(m.name)
```

```python
def story_teller(topic,genre,length=200):
  prompt=f"""Write a creative story based on topic,genre and length
          topic:{topic},
          genre:{genre},
          length:{length}

          Give story for kids dont give high end english story
          """
  response=client.models.generate_content(
      model='gemini-3.5-flash-lite',
      contents=prompt,
      config=types.GenerateContentConfig(
          temperature=0.6,
          max_output_tokens=length,

      )
  )
  return response.text
topic=input("Enter the topic")
genre=input("Enter the genre")
length=int(input("Enter the length"))
story=story_teller(topic,genre,length)
print(story)
```
