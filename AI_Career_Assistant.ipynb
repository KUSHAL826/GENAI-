```python
from google import genai
from google.colab import userdata
from google.genai import types

client=genai.Client(api_key=userdata.get("GEMINI_API_KEYS"))
MODEL="gemini-3.5-flash-lite"
```

```python
system_instruction="""
User enters with this interested Domain for user further studies, the user also enters with current education with their result.
analyze interested domain, current education and result and provide career roadmap for user in structured text format with explination
if current education result is to low for percieving further studies in interested domain suggest other domain with explination.
if current education result is more for precieving further studies and have chance of percieving better domain suggest the same

Provide after 2-3times validation
Reply whenever new session starts reply with (hi,i am ur career support assistant)
"""

chat=client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        temperature=0.4,
        max_output_tokens=2000,
        system_instruction=system_instruction,
        thinking_config=types.ThinkingConfig(thinking_level='medium')
    )
)
print("Enter exit,bye,quit to exit session\n")
while True:
  user_input=input("YOU: ")
  if user_input.lower() in ['exit','quit','bye']:
    print("ALL THE BEST FOR YOUR FURTHER CAREER\n")
    break
  response=chat.send_message(user_input)
  print(f"CAREER ASSISTANT: {response.text}")
```
